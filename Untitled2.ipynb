```python
import pandas as pd

try:
    # Read the CSV without a header and specify the delimiter
    df = pd.read_csv("/content/EURUSD240.csv", header=None, sep='\t')
    display(df.head())
except FileNotFoundError:
    print("Error: Make sure the file '/content/EURUSD240.csv' exists in your Colab environment.")
```

```python
import pandas as pd
import numpy as np

# Assuming the DataFrame 'df' is already loaded from the previous step and has no header

# Rename the columns
df.columns = ['Date', 'Open', 'High', 'Low', 'Close', 'Volume']

# Convert relevant columns to numeric, coercing errors
df[['Open', 'High', 'Low', 'Close', 'Volume']] = df[['Open', 'High', 'Low', 'Close', 'Volume']].apply(pd.to_numeric, errors='coerce')

# Convert 'Date' column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Set 'Date' as the index
df.set_index('Date', inplace=True)

# Drop rows with NaN values that resulted from errors='coerce'
df.dropna(inplace=True)


# Calculate Williams Fractals
def williams_fractals(df, n=2):
    df['Fractal_High'] = np.nan
    df['Fractal_Low'] = np.nan

    # Ensure there are enough data points to calculate fractals
    if len(df) > 2 * n:
        for i in range(n, len(df) - n):
            # Check for bullish fractal (high)
            if df['High'].iloc[i] > df['High'].iloc[i-n:i].max() and \
               df['High'].iloc[i] > df['High'].iloc[i+1:i+n+1].max():
                df.loc[df.index[i], 'Fractal_High'] = df['High'].iloc[i]

            # Check for bearish fractal (low)
            if df['Low'].iloc[i] < df['Low'].iloc[i-n:i].min() and \
               df['Low'].iloc[i] < df['Low'].iloc[i+1:i+n+1].min():
                df.loc[df.index[i], 'Fractal_Low'] = df['Low'].iloc[i]
    return df

df = williams_fractals(df)

# Calculate Bollinger Bands
window = 20
num_std_dev = 2
df['rolling_mean'] = df['Close'].rolling(window=window).mean()
df['rolling_std'] = df['Close'].rolling(window=window).std()
df['Upper_Band'] = df['rolling_mean'] + (df['rolling_std'] * num_std_dev)
df['Lower_Band'] = df['rolling_mean'] - (df['rolling_std'] * num_std_dev)

# Calculate Bollinger Bands %b
df['Percent_B'] = (df['Close'] - df['Lower_Band']) / (df['Upper_Band'] - df['Lower_Band'])

# Display the DataFrame with new indicator columns
display(df.head())
display(df.tail())
```

```python
# Define trading strategy based on indicators

# Initialize columns for buy and sell signals
df['Buy_Signal'] = False
df['Sell_Signal'] = False

# Strategy rules: Fractal confirmation and Bollinger Bands conditions.

confirm_bars = 1 # Require 1 bar confirmation after the fractal

# Iterate through the DataFrame to find signals based on the new rules
# We need at least n+confirm_bars bars to check for a confirmed fractal
n = 2 # Assuming n=2 from the fractal calculation

for i in range(n + confirm_bars, len(df)):
    # Check for Buy signal (based on a confirmed Bearish Fractal - low)
    # A bearish fractal occurred 'confirm_bars' bars ago, and the price
    # has stayed above that fractal low for 'confirm_bars' bars.
    # Also, check Bollinger Bands %b for oversold conditions.
    if (df['Fractal_Low'].iloc[i - confirm_bars] > 0 and
        all(df['Low'].iloc[i - j] > df['Fractal_Low'].iloc[i - confirm_bars] for j in range(confirm_bars)) and
        df['Percent_B'].iloc[i] < 0.2): # Example Bollinger Band condition
        df.loc[df.index[i], 'Buy_Signal'] = True

    # Check for Sell signal (based on a confirmed Bullish Fractal - high)
    # A bullish fractal occurred 'confirm_bars' bars ago, and the price
    # has stayed below that fractal high for 'confirm_bars' bars.
    # Also, check Bollinger Bands %b for overbought conditions.
    elif (df['Fractal_High'].iloc[i - confirm_bars] > 0 and
          all(df['High'].iloc[i - j] < df['Fractal_High'].iloc[i - confirm_bars] for j in range(confirm_bars)) and
          df['Percent_B'].iloc[i] > 0.8): # Example Bollinger Band condition
        df.loc[df.index[i], 'Sell_Signal'] = True


# Display the DataFrame with signals
display(df.head())
display(df.tail())
```

```python
# Backtesting the trading strategy

initial_balance = 10000  # Starting capital
balance = initial_balance
position = 0  # 0 for no position, 1 for long, -1 for short
trades = []
entry_price = 0
stop_loss = 0
take_profit = 0
pip_value = 10 # For EUR/USD with a standard lot of 100,000 units, the value of one pip is $10.
pip_decimal_places = 5 # EUR/USD typically has 5 decimal places

for i in range(len(df)):
    # Get current bar information
    date = df.index[i]
    open_price = df['Open'].iloc[i]
    high_price = df['High'].iloc[i]
    low_price = df['Low'].iloc[i]
    close_price = df['Close'].iloc[i]
    buy_signal = df['Buy_Signal'].iloc[i]
    sell_signal = df['Sell_Signal'].iloc[i]

    # Check for exits first (Stop Loss or Take Profit)
    if position == 1: # Long position
        # Check for Stop Loss
        if low_price <= stop_loss:
            exit_price = stop_loss
            profit_pips = (exit_price - entry_price) * (10 ** pip_decimal_places)
            profit_usd = profit_pips * pip_value
            balance += profit_usd
            trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Long', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Stop Loss'})
            position = 0
            entry_price = 0
            stop_loss = 0
            take_profit = 0
            continue # Move to the next bar after exiting

        # Check for Take Profit
        if high_price >= take_profit:
            exit_price = take_profit
            profit_pips = (exit_price - entry_price) * (10 ** pip_decimal_places)
            profit_usd = profit_pips * pip_value
            balance += profit_usd
            trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Long', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Take Profit'})
            position = 0
            entry_price = 0
            stop_loss = 0
            take_profit = 0
            continue # Move to the next bar after exiting

    elif position == -1: # Short position
        # Check for Stop Loss
        if high_price >= stop_loss:
            exit_price = stop_loss
            profit_pips = (entry_price - exit_price) * (10 ** pip_decimal_places)
            profit_usd = profit_pips * pip_value
            balance += profit_usd
            trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Short', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Stop Loss'})
            position = 0
            entry_price = 0
            stop_loss = 0
            take_profit = 0
            continue # Move to the next bar after exiting

        # Check for Take Profit
        if low_price <= take_profit:
            exit_price = take_profit
            profit_pips = (entry_price - exit_price) * (10 ** pip_decimal_places)
            profit_usd = profit_pips * pip_value
            balance += profit_usd
            trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Short', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Take Profit'})
            position = 0
            entry_price = 0
            stop_loss = 0
            take_profit = 0
            continue # Move to the next bar after exiting


    # Check for new entry signals if no position is currently open
    if position == 0:
        # Buy signal
        if buy_signal:
            entry_date = date
            entry_price = close_price
            position = 1 # Enter long position
            # Define Stop Loss and Take Profit for long position (example - adjust as needed)
            stop_loss = entry_price - (50 * (10**-pip_decimal_places)) # 50 pips stop loss
            take_profit = entry_price + (100 * (10**-pip_decimal_places)) # 100 pips take profit
            # Adjust stop loss and take profit to be within reasonable bounds of the bar
            # (This is a simplification; a real backtest would handle this more carefully)
            if stop_loss < low_price: stop_loss = low_price
            if take_profit > high_price: take_profit = high_price


        # Sell signal
        elif sell_signal:
            entry_date = date
            entry_price = close_price
            position = -1 # Enter short position
            # Define Stop Loss and Take Profit for short position (example - adjust as needed)
            stop_loss = entry_price + (50 * (10**-pip_decimal_places)) # 50 pips stop loss
            take_profit = entry_price - (100 * (10**-pip_decimal_places)) # 100 pips take profit
             # Adjust stop loss and take profit to be within reasonable bounds of the bar
            # (This is a simplification; a real backtest would handle this more carefully)
            if stop_loss > high_price: stop_loss = high_price
            if take_profit < low_price: take_profit = low_price


# If a position is still open at the end of the data, close it at the last closing price
if position != 0:
    exit_price = df['Close'].iloc[-1]
    profit_pips = (exit_price - entry_price) * (10 ** pip_decimal_places) if position == 1 else (entry_price - exit_price) * (10 ** pip_decimal_places)
    profit_usd = profit_pips * pip_value
    balance += profit_usd
    trades.append({'Entry Date': entry_date, 'Exit Date': df.index[-1], 'Type': 'Long' if position == 1 else 'Short', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'End of Data'})

# Convert trades list to a DataFrame for easier analysis
trades_df = pd.DataFrame(trades)

# Display backtesting results
print("Backtesting Complete.")
print(f"Initial Balance: ${initial_balance:.2f}")
print(f"Final Balance: ${balance:.2f}")
print(f"Total Profit/Loss: ${balance - initial_balance:.2f}")
print(f"Number of Trades: {len(trades_df)}")

if not trades_df.empty:
    print("\nTrades:")
    display(trades_df.head())
    display(trades_df.tail())

    # Basic performance metrics
    winning_trades = trades_df[trades_df['Profit (USD)'] > 0]
    losing_trades = trades_df[trades_df['Profit (USD)'] < 0]

    print(f"\nNumber of Winning Trades: {len(winning_trades)}")
    print(f"Number of Losing Trades: {len(losing_trades)}")
    if len(trades_df) > 0:
        print(f"Win Rate: {len(winning_trades) / len(trades_df) * 100:.2f}%")

    if not winning_trades.empty:
        print(f"Average Win: ${winning_trades['Profit (USD)'].mean():.2f}")
    if not losing_trades.empty:
        print(f"Average Loss: ${losing_trades['Profit (USD)'].mean():.2f}")

    # Calculate Drawdown (simplified)
    # equity_curve = trades_df['Profit (USD)'].cumsum() + initial_balance
    # peak = equity_curve.expanding(min_periods=1).max()
    # drawdown = (equity_curve - peak) / peak
    # max_drawdown = drawdown.min()
    # print(f"Maximum Drawdown: {max_drawdown:.2%}")
```

## Modify backtesting for scaled profit target exit

### Subtask:
Adapt the backtesting code to:
* Enter trades based on the confirmed signals.
* Execute the partial exit at 10 or 20 pips profit (as before).
* Close the remaining half of the position when a specified profit target (from the defined range) is reached.
* Track the profit/loss for both parts of the trade.


**Reasoning**:
Define the `backtest_strategy` function with the specified parameters and implement the backtesting logic, including entry, partial exit, scaled profit target exit, and opposite fractal exit.



```python
import pandas as pd
import numpy as np

def backtest_strategy(df, partial_exit_pips, scaled_profit_target_pips, stop_loss_usd=None):
    """
    Backtests the trading strategy with partial take profit and scaled profit target,
    with an optional fixed USD stop loss for the second half of the position.

    Args:
        df: pandas DataFrame with trading data and signals.
        partial_exit_pips: The profit target in pips for the first half of the position.
        scaled_profit_target_pips: The profit target in pips for the second half of the position.
        stop_loss_usd: Optional. A fixed USD amount for the stop loss on the second half
                       of the position. If None, the opposite fractal exit is used.

    Returns:
        A tuple containing:
            trades_df: pandas DataFrame with details of each trade.
            balance: The final balance after backtesting.
    """
    initial_balance = 10000  # Starting capital
    balance = initial_balance
    position = 0  # 0 for no position, 1 for long, -1 for short
    trades = []
    entry_price = 0
    entry_date = None
    position_size = 0 # 1 for full position, 0.5 for half position
    pip_value = 10 # For EUR/USD with a standard lot of 100,000 units, the value of one pip is $10.
    pip_decimal_places = 5 # EUR/USD typically has 5 decimal places

    # Function to calculate pips
    def calculate_pips(entry, exit, trade_type):
        if trade_type == 'Long':
            return (exit - entry) * (10 ** pip_decimal_places)
        elif trade_type == 'Short':
            return (entry - exit) * (10 ** pip_decimal_places)
        return 0

    # Function to calculate price based on pips
    def calculate_price_from_pips(entry, pips, trade_type):
        if trade_type == 'Long':
            return entry + (pips * (10**-pip_decimal_places))
        elif trade_type == 'Short':
            return entry - (pips * (10**-pip_decimal_places))
        return entry


    for i in range(len(df)):
        # Get current bar information
        date = df.index[i]
        open_price = df['Open'].iloc[i]
        high_price = df['High'].iloc[i]
        low_price = df['Low'].iloc[i]
        close_price = df['Close'].iloc[i]
        buy_signal = df['Buy_Signal'].iloc[i]
        sell_signal = df['Sell_Signal'].iloc[i]
        fractal_high = df['Fractal_High'].iloc[i]
        fractal_low = df['Fractal_Low'].iloc[i]


        # --- Exit Logic ---
        if position != 0:
            # Check for Partial Take Profit
            if position_size == 1: # Check if the full position is open
                if (position == 1 and high_price >= entry_price + (partial_exit_pips * (10**-pip_decimal_places))) or \
                   (position == -1 and low_price <= entry_price - (partial_exit_pips * (10**-pip_decimal_places))):

                    partial_exit_price = entry_price + (partial_exit_pips * (10**-pip_decimal_places)) if position == 1 else entry_price - (partial_exit_pips * (10**-pip_decimal_places))
                    profit_pips = calculate_pips(entry_price, partial_exit_price, 'Long' if position == 1 else 'Short')
                    profit_usd = profit_pips * pip_value * 0.5 # Half the position size
                    balance += profit_usd
                    trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Long' if position == 1 else 'Short', 'Entry Price': entry_price, 'Exit Price': partial_exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Partial Take Profit'})
                    position_size = 0.5 # Half position remaining

            # Check for Scaled Profit Target Exit (for the remaining half position)
            if position_size == 0.5: # Check if half position is open
                scaled_target_price = entry_price + (scaled_profit_target_pips * (10**-pip_decimal_places)) if position == 1 else entry_price - (scaled_profit_target_pips * (10**-pip_decimal_places))
                if (position == 1 and high_price >= scaled_target_price) or \
                   (position == -1 and low_price <= scaled_target_price):

                    exit_price = scaled_target_price
                    profit_pips = calculate_pips(entry_price, exit_price, 'Long' if position == 1 else 'Short')
                    profit_usd = profit_pips * pip_value * 0.5 # Remaining half position
                    balance += profit_usd
                    trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Long' if position == 1 else 'Short', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Scaled Profit Target'})
                    position = 0
                    position_size = 0
                    entry_price = 0
                    entry_date = None # Reset entry date
                    continue # Move to the next bar after exiting the full position

            # Check for Fixed USD Stop Loss (for the remaining half position)
            if position_size == 0.5 and stop_loss_usd is not None:
                stop_loss_pips = stop_loss_usd / (pip_value * 0.5) # Calculate equivalent pips for half position
                stop_loss_price = calculate_price_from_pips(entry_price, -stop_loss_pips, 'Long' if position == 1 else 'Short') # Negative pips for loss


                if (position == 1 and low_price <= stop_loss_price) or \
                   (position == -1 and high_price >= stop_loss_price):

                    exit_price = stop_loss_price
                    profit_pips = calculate_pips(entry_price, exit_price, 'Long' if position == 1 else 'Short')
                    profit_usd = profit_pips * pip_value * 0.5 # Remaining half position
                    balance += profit_usd
                    trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Long' if position == 1 else 'Short', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Fixed USD Stop Loss'})
                    position = 0
                    position_size = 0
                    entry_price = 0
                    entry_date = None # Reset entry date
                    continue # Move to the next bar after exiting the full position


            # Check for Exit (Opposite Fractal) for the remaining half position if not already exited
            if position_size == 0.5 and stop_loss_usd is None: # Only use fractal exit if no fixed stop loss is set
                 if (position == 1 and fractal_high > 0) or \
                    (position == -1 and fractal_low > 0):

                    exit_price = close_price # Exit at the close of the fractal candle
                    profit_pips = calculate_pips(entry_price, exit_price, 'Long' if position == 1 else 'Short')
                    profit_usd = profit_pips * pip_value * 0.5 # Remaining half position
                    balance += profit_usd
                    trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Long' if position == 1 else 'Short', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Opposite Fractal'})
                    position = 0
                    position_size = 0
                    entry_price = 0
                    entry_date = None # Reset entry date
                    continue # Move to the next bar after exiting the full position


        # --- Entry Logic ---
        if position == 0:
            # Buy signal
            if buy_signal:
                entry_date = date
                entry_price = close_price
                position = 1 # Enter long position
                position_size = 1 # Full position size


            # Sell signal
            elif sell_signal:
                entry_date = date
                entry_price = close_price
                position = -1 # Enter short position
                position_size = 1 # Full position size


    # If a position is still open at the end of the data, close it at the last closing price
    if position != 0:
        exit_price = df['Close'].iloc[-1]
        profit_pips = calculate_pips(entry_price, exit_price, 'Long' if position == 1 else 'Short')
        profit_usd = profit_pips * pip_value * position_size # Account for remaining position size
        balance += profit_usd
        trades.append({'Entry Date': entry_date, 'Exit Date': df.index[-1], 'Type': 'Long' if position == 1 else 'Short', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'End of Data'})


    # Convert trades list to a DataFrame for easier analysis
    trades_df = pd.DataFrame(trades)

    return trades_df, balance

# Example usage (you will run this in the next subtask for different profit targets)
# trades_df, final_balance = backtest_strategy(df.copy(), partial_exit_pips=20, scaled_profit_target_pips=100)
# print(f"Final Balance: ${final_balance:.2f}")
# display(trades_df.head())
# display(trades_df.tail())
```

```python
# Initialize a dictionary to store backtesting results for each profit target
results = {}

# Choose the partial exit pips (either 10 or 20)
chosen_partial_exit_pips = 20

# Iterate through the defined range of profit targets
for profit_target in profit_target_range:
    print(f"Backtesting with Partial Exit: {chosen_partial_exit_pips} pips, Scaled Profit Target: {profit_target} pips")
    # Run the backtest for the current profit target
    trades_df_result, final_balance = backtest_strategy(df.copy(), partial_exit_pips=chosen_partial_exit_pips, scaled_profit_target_pips=profit_target)

    # Store the results
    results[profit_target] = {
        'final_balance': final_balance,
        'total_profit_loss': final_balance - 10000, # Assuming initial balance is 10000
        'number_of_trades': len(trades_df_result),
        'winning_trades_count': len(trades_df_result[trades_df_result['Profit (USD)'] > 0]) if not trades_df_result.empty else 0,
        'losing_trades_count': len(trades_df_result[trades_df_result['Profit (USD)'] < 0]) if not trades_df_result.empty else 0
    }
    print(f"Finished backtesting for {profit_target} pips. Final Balance: ${final_balance:.2f}")

# Print or display the collected results
print("\nBacktesting Results Summary:")
for profit_target, data in results.items():
    print(f"Profit Target: {profit_target} pips")
    print(f"  Final Balance: ${data['final_balance']:.2f}")
    print(f"  Total Profit/Loss: ${data['total_profit_loss']:.2f}")
    print(f"  Number of Trades: {data['number_of_trades']}")
    print(f"  Winning Trades: {data['winning_trades_count']}")
    print(f"  Losing Trades: {data['losing_trades_count']}")
    if data['number_of_trades'] > 0:
        print(f"  Win Rate: {data['winning_trades_count'] / data['number_of_trades'] * 100:.2f}%")
    print("-" * 20)
```

```python
import pandas as pd
import numpy as np

def backtest_strategy(df, partial_exit_pips, scaled_profit_target_pips, stop_loss_usd=None):
    """
    Backtests the trading strategy with partial take profit and scaled profit target,
    with an optional fixed USD stop loss for the second half of the position.

    Args:
        df: pandas DataFrame with trading data and signals.
        partial_exit_pips: The profit target in pips for the first half of the position.
        scaled_profit_target_pips: The profit target in pips for the second half of the position.
        stop_loss_usd: Optional. A fixed USD amount for the stop loss on the second half
                       of the position. If None, the opposite fractal exit is used.

    Returns:
        A tuple containing:
            trades_df: pandas DataFrame with details of each trade.
            balance: The final balance after backtesting.
    """
    initial_balance = 10000  # Starting capital
    balance = initial_balance
    position = 0  # 0 for no position, 1 for long, -1 for short
    trades = []
    entry_price = 0
    entry_date = None
    position_size = 0 # 1 for full position, 0.5 for half position
    pip_value = 10 # For EUR/USD with a standard lot of 100,000 units, the value of one pip is $10.
    pip_decimal_places = 5 # EUR/USD typically has 5 decimal places

    # Function to calculate pips
    def calculate_pips(entry, exit, trade_type):
        if trade_type == 'Long':
            return (exit - entry) * (10 ** pip_decimal_places)
        elif trade_type == 'Short':
            return (entry - exit) * (10 ** pip_decimal_places)
        return 0

    # Function to calculate price based on pips
    def calculate_price_from_pips(entry, pips, trade_type):
        if trade_type == 'Long':
            return entry + (pips * (10**-pip_decimal_places))
        elif trade_type == 'Short':
            return entry - (pips * (10**-pip_decimal_places))
        return entry


    for i in range(len(df)):
        # Get current bar information
        date = df.index[i]
        open_price = df['Open'].iloc[i]
        high_price = df['High'].iloc[i]
        low_price = df['Low'].iloc[i]
        close_price = df['Close'].iloc[i]
        buy_signal = df['Buy_Signal'].iloc[i]
        sell_signal = df['Sell_Signal'].iloc[i]
        fractal_high = df['Fractal_High'].iloc[i]
        fractal_low = df['Fractal_Low'].iloc[i]


        # --- Exit Logic ---
        if position != 0:
            # Check for Partial Take Profit
            if position_size == 1: # Check if the full position is open
                if (position == 1 and high_price >= entry_price + (partial_exit_pips * (10**-pip_decimal_places))) or \
                   (position == -1 and low_price <= entry_price - (partial_exit_pips * (10**-pip_decimal_places))):

                    partial_exit_price = entry_price + (partial_exit_pips * (10**-pip_decimal_places)) if position == 1 else entry_price - (partial_exit_pips * (10**-pip_decimal_places))
                    profit_pips = calculate_pips(entry_price, partial_exit_price, 'Long' if position == 1 else 'Short')
                    profit_usd = profit_pips * pip_value * 0.5 # Half the position size
                    balance += profit_usd
                    trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Long' if position == 1 else 'Short', 'Entry Price': entry_price, 'Exit Price': partial_exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Partial Take Profit'})
                    position_size = 0.5 # Half position remaining

            # Check for Scaled Profit Target Exit (for the remaining half position)
            if position_size == 0.5: # Check if half position is open
                scaled_target_price = entry_price + (scaled_profit_target_pips * (10**-pip_decimal_places)) if position == 1 else entry_price - (scaled_profit_target_pips * (10**-pip_decimal_places))
                if (position == 1 and high_price >= scaled_target_price) or \
                   (position == -1 and low_price <= scaled_target_price):

                    exit_price = scaled_target_price
                    profit_pips = calculate_pips(entry_price, exit_price, 'Long' if position == 1 else 'Short')
                    profit_usd = profit_pips * pip_value * 0.5 # Remaining half position
                    balance += profit_usd
                    trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Long' if position == 1 else 'Short', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Scaled Profit Target'})
                    position = 0
                    position_size = 0
                    entry_price = 0
                    entry_date = None # Reset entry date
                    continue # Move to the next bar after exiting the full position

            # Check for Fixed USD Stop Loss (for the remaining half position)
            if position_size == 0.5 and stop_loss_usd is not None:
                stop_loss_pips = stop_loss_usd / (pip_value * 0.5) # Calculate equivalent pips for half position
                stop_loss_price = calculate_price_from_pips(entry_price, -stop_loss_pips, 'Long' if position == 1 else 'Short') # Negative pips for loss


                if (position == 1 and low_price <= stop_loss_price) or \
                   (position == -1 and high_price >= stop_loss_price):

                    exit_price = stop_loss_price
                    profit_pips = calculate_pips(entry_price, exit_price, 'Long' if position == 1 else 'Short')
                    profit_usd = profit_pips * pip_value * 0.5 # Remaining half position
                    balance += profit_usd
                    trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Long' if position == 1 else 'Short', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Fixed USD Stop Loss'})
                    position = 0
                    position_size = 0
                    entry_price = 0
                    entry_date = None # Reset entry date
                    continue # Move to the next bar after exiting the full position


            # Check for Exit (Opposite Fractal) for the remaining half position if not already exited
            if position_size == 0.5 and stop_loss_usd is None: # Only use fractal exit if no fixed stop loss is set
                 if (position == 1 and fractal_high > 0) or \
                    (position == -1 and fractal_low > 0):

                    exit_price = close_price # Exit at the close of the fractal candle
                    profit_pips = calculate_pips(entry_price, exit_price, 'Long' if position == 1 else 'Short')
                    profit_usd = profit_pips * pip_value * 0.5 # Remaining half position
                    balance += profit_usd
                    trades.append({'Entry Date': entry_date, 'Exit Date': date, 'Type': 'Long' if position == 1 else 'Short', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'Opposite Fractal'})
                    position = 0
                    position_size = 0
                    entry_price = 0
                    entry_date = None # Reset entry date
                    continue # Move to the next bar after exiting the full position


        # --- Entry Logic ---
        if position == 0:
            # Buy signal
            if buy_signal:
                entry_date = date
                entry_price = close_price
                position = 1 # Enter long position
                position_size = 1 # Full position size


            # Sell signal
            elif sell_signal:
                entry_date = date
                entry_price = close_price
                position = -1 # Enter short position
                position_size = 1 # Full position size


    # If a position is still open at the end of the data, close it at the last closing price
    if position != 0:
        exit_price = df['Close'].iloc[-1]
        profit_pips = calculate_pips(entry_price, exit_price, 'Long' if position == 1 else 'Short')
        profit_usd = profit_pips * pip_value * position_size # Account for remaining position size
        balance += profit_usd
        trades.append({'Entry Date': entry_date, 'Exit Date': df.index[-1], 'Type': 'Long' if position == 1 else 'Short', 'Entry Price': entry_price, 'Exit Price': exit_price, 'Profit (pips)': profit_pips, 'Profit (USD)': profit_usd, 'Exit Reason': 'End of Data'})


    # Convert trades list to a DataFrame for easier analysis
    trades_df = pd.DataFrame(trades)

    return trades_df, balance

# Backtest the strategy on M30 data using the defined backtest_strategy function
# Use the optimal scaled profit target found during the H4 analysis (300 pips) and 20 pips partial exit
partial_exit_pips_m30 = 20
scaled_profit_target_pips_m30 = 300 # Using the optimal target from H4 analysis

print(f"Backtesting strategy on M30 data with Partial Exit: {partial_exit_pips_m30} pips, Scaled Profit Target: {scaled_profit_target_pips_m30} pips")
trades_df_m30, final_balance_m30 = backtest_strategy(df_m30.copy(), partial_exit_pips=partial_exit_pips_m30, scaled_profit_target_pips=scaled_profit_target_pips_m30)

# Display backtesting results for M30
print("\nBacktesting Complete (M30 Data).")
print(f"Initial Balance: ${10000:.2f}") # Assuming initial balance is consistent
print(f"Final Balance: ${final_balance_m30:.2f}")
print(f"Total Profit/Loss: ${final_balance_m30 - 10000:.2f}")
print(f"Number of Trades: {len(trades_df_m30)}")

if not trades_df_m30.empty:
    print("\nTrades (M30 Data):")
    display(trades_df_m30.head())
    display(trades_df_m30.tail())

    # Basic performance metrics for M30
    winning_trades_m30 = trades_df_m30[trades_df_m30['Profit (USD)'] > 0]
    losing_trades_m30 = trades_df_m30[trades_df_m30['Profit (USD)'] < 0]

    print(f"\nNumber of Winning Trades (M30): {len(winning_trades_m30)}")
    print(f"Number of Losing Trades (M30): {len(losing_trades_m30)}")
    if len(trades_df_m30) > 0:
        print(f"Win Rate (M30): {len(winning_trades_m30) / len(trades_df_m30) * 100:.2f}%")

    if not winning_trades_m30.empty:
        print(f"Average Win (M30): ${winning_trades_m30['Profit (USD)'].mean():.2f}")
    if not losing_trades_m30.empty:
        print(f"Average Loss (M30): ${losing_trades_m30['Profit (USD)'].mean():.2f}")
```
